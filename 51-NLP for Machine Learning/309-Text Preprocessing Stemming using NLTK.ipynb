### PorterStemmer


```python
## Comments of product is a positive review or negative review. The comments are in the form of a list of strings. The function should return a list of strings where each string is either "positive" or "negative" depending on the sentiment of the corresponding comment.
## Reviews --> eating,eats,eaten[going,gonee,goes]--> go

words = ["eating", "eats", "eaten","writing", "write","programming","programs","history","finally","finalized"]
```

```python
pip install nltk
```

```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
for word in words:
    print(word, " --> ", stemming.stem(word))
```

```python
stemming.stem("Congratulations")
```

```python
stemming.stem("sitting")
```

### RegexStemmer class


```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|ed$', min=4)
```

```python
reg_stemmer.stem("eating")
```

```python

reg_stemmer.stem("ingeating")
```

### Snowball Stemmer


```python
from nltk.stem import SnowballStemmer
snowballstemmer = SnowballStemmer("english")
for word in words:
    print(word, " --> ", snowballstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballstemmer.stem("fairly"),snowballstemmer.stem("sportingly")
```
