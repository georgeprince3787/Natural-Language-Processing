```python
corpus ="""Hello Welcome to the world of Natural Language Processing.
In this practical, we will learn about tokenization, which is the process of breaking down text into smaller units called tokens. 
Tokens can be words, phrases, or even sentences. 
Tokenization is a crucial step in NLP as it helps in understanding the structure and meaning of the text.
"""
```

```python
print(corpus)
```

```python
### Tokenization using NLTK
## Sentence-->paragrapghs
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
### Tokenization
## paragraph-->words
## sentence-->words
from nltk.tokenize import word_tokenize

words = word_tokenize(corpus)
print(words)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
